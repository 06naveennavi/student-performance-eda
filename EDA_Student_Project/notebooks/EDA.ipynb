```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/StudentsPerformance.csv')
df.head()
```

```python
df.shape
```

```python
df.columns
```

```python
df.info()
```

```python
df.columns = df.columns.str.replace(' ', '_')
df.columns
```

```python
df.describe()
```

```python
sns.histplot(df['math_score'], kde=True)
plt.title("Distribution of Math Scores")
plt.show()
```

```python
sns.histplot(df['reading_score'], kde=True)
plt.title("Reading Score Distribution")
plt.show()

sns.histplot(df['writing_score'], kde=True)
plt.title("Writing Score Distribution")
plt.show()
```

```python
sns.countplot(x='gender', data=df)
plt.title("Gender Count")
plt.show()
```

```python
sns.boxplot(x='test_preparation_course', y='math_score', data=df)
plt.title("Test Preparation vs Math Score")
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
df.groupby('gender')[['math_score','reading_score','writing_score']].mean()
```

```python
df.groupby('test_preparation_course')[['math_score']].mean()
```

```python
df.groupby('parental_level_of_education')[['math_score']].mean().sort_values(by='math_score')
```

##key insights from EDA:
- Students who completed the test preparation course scored significantly higher in math compared to those who did not.

- Reading and writing scores show a strong positive correlation, indicating that students who perform well in reading also tend to perform well in writing.

- Female students tend to perform better in reading and writing, while math scores are relatively similar between genders.

- Students whose parents have higher levels of education tend to score better, suggesting a possible influence of family educational background on student performance.

- Most students have average scores, with fewer students achieving extremely high or low marks, indicating a roughly normal distribution.
